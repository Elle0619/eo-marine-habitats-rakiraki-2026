### 1. Enter the path to your output classification data

```python
# Path to your GeoTIFF raster file
raster_path = "nm-mana.tif"
```

### 2. Enter the class code you are interested in:

```python
# Target pixel value to count (e.g., class code 1)
target_class = 1
```

### 3. Open the file

```python
# Open the raster file
with rasterio.open(raster_path) as src:
    # Read the first band as a NumPy array (change band index if multi-band)
    band1 = src.read(target_class)
```

### 4. Count the pixels

```python
# Count the number of pixels matching target_value
pixel_count = np.sum(band1 == target_class)

print(f"Number of pixels with value {target_class}: {pixel_count:,}")
```
